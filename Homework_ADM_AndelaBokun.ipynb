```python
#Text Wrap
import textwrap

def wrap(string, max_width):
    def1 = max_width
    s1 = []
    i1=0
    while(i1<=len(string)):
        s1.append(string[i1:max_width])
        i1+=def1
        max_width += def1

    s2 = "\n".join(s1)
    return s2

if __name__ == '__main__':
    string, max_width = input(), int(input())
    result = wrap(string, max_width)
    print(result)
```

```python
#String Formatting
def print_formatted(number):
    decimalni = []
    for i in range(1, number+1):
        decimalni.append(i)

    binarni = []
    for i in range(1, number+1):
        bin_verzija = bin(i)[2:]
        binarni.append(bin_verzija)

    octal1 = []
    for i in range(1, number+1):
        octal_verzija = oct(i)[2:]
        octal1.append(octal_verzija)

    hex1 = []
    for i in range(1, number+1):
        hex_verzija = hex(i)[2:].upper()
        hex1.append(hex_verzija)

    sirina = len(binarni[-1])

    for k in range(number):
        print(str(decimalni[k]).rjust(sirina), octal1[k].rjust(sirina), hex1[k].rjust(sirina), binarni[k].rjust(sirina))



if __name__ == '__main__':
    n = int(input())
    print_formatted(n)
```

```python
#Alphabet Rangoli
import string
def print_rangoli(size):
    alfabet = string.ascii_lowercase
    slova = []
    for i in range(0, size):
        slova.append(alfabet[i])

    slova.sort(reverse=True)

    gore = []
    for i in range(0, size):
        levaSlova = list(slova[:i+1])
        desnaSlova = levaSlova[:-1][::-1]
        red = "-".join(levaSlova+desnaSlova)
        gore.append(red.center(4*size - 3, '-'))

    dole = gore[-2::-1]

    print("\n".join(gore+dole))


if __name__ == '__main__':
    n = int(input())
    print_rangoli(n)
```

```python
#Time Delta
#!/bin/python3

import math
import os
import random
import re
import sys
from datetime import datetime

# Complete the time_delta function below.
def time_delta(t1, t2):
    fmt = '%a %d %b %Y %H:%M:%S %z'

    tt1 = datetime.strptime(t1, fmt)
    tt2 = datetime.strptime(t2, fmt)
    razlika = abs((tt1 - tt2).total_seconds())
    return str(int(razlika))

if __name__ == '__main__':
    fptr = open(os.environ['OUTPUT_PATH'], 'w')

    t = int(input())

    for t_itr in range(t):
        t1 = input()

        t2 = input()

        delta = time_delta(t1, t2)

        fptr.write(delta + '\n')

    fptr.close()
```

```python
#Viral Advertising
#!/bin/python3

import math
import os
import random
import re
import sys

#
# Complete the 'viralAdvertising' function below.
#
# The function is expected to return an INTEGER.
# The function accepts INTEGER n as parameter.
#

def viralAdvertising(n):
    shared = 5
    cumulative = 0
    for i in range(1, n+1):
        liked = math.floor(shared/2)
        cumulative+=liked
        shared = 3*liked
    return cumulative

if __name__ == '__main__':
    fptr = open(os.environ['OUTPUT_PATH'], 'w')

    n = int(input().strip())

    result = viralAdvertising(n)

    fptr.write(str(result) + '\n')

    fptr.close()
```

```python
#XML1 Find the score
import sys
import xml.etree.ElementTree as etree

def get_attr_number(node):
    s1 = len(node.attrib)
    for child in node:
        s1 += get_attr_number(child)
    return s1

if __name__ == '__main__':
    sys.stdin.readline()
    xml = sys.stdin.read()
    tree = etree.ElementTree(etree.fromstring(xml))
    root = tree.getroot()
    print(get_attr_number(root))
```

```python
#XML2 Find the Maximum Depth
import xml.etree.ElementTree as etree

maxdepth = 0
def depth(elem, level):
    global maxdepth
    level +=1
    if maxdepth < level:
        maxdepth = level
    for child in elem:
        depth(child, level)

if __name__ == '__main__':
    n = int(input())
    xml = ""
    for i in range(n):
        xml =  xml + input() + "\n"
    tree = etree.ElementTree(etree.fromstring(xml))
    depth(tree.getroot(), -1)
    print(maxdepth)
```

```python
#Minion Game
def minion_game(string):
    vowels = "AEIOU"
    n = len(string)
    kevin = 0
    stuart = 0

    for i in range(n):
        if string[i] in vowels:
            kevin += n - i
        else:
            stuart += n - i

    if kevin > stuart:
        print("Kevin", kevin)
    elif stuart > kevin:
        print("Stuart", stuart)
    else:
        print("Draw")


if __name__ == '__main__':
    s = input()
    minion_game(s)
```

```python
#HTML Parser Part1
from html.parser import HTMLParser

class MyHTMLParser(HTMLParser):
    def handle_starttag(self, tag, attrs):
        print("Start :", tag)
        for ime, vrednost in attrs:
            print("->", ime, ">", vrednost)

    def handle_endtag(self, tag):
        print("End   :", tag)

    def handle_startendtag(self, tag, attrs):
        print("Empty :", tag)
        for ime, vrednost in attrs:
            print("->", ime, ">", vrednost)

N = int(input())
parser = MyHTMLParser()

for _ in range(N):
    parser.feed(input())
```

```python
#Collections.deque()
from collections import deque

d = deque()
N = int(input())

for _ in range(N):
    kom = input().split()
    if kom[0] == "append":
        d.append(kom[1])
    if kom[0] == "pop":
        d.pop()
    if kom[0] == "popleft":
        d.popleft()
    if kom[0] == "appendleft":
        d.appendleft(kom[1])

for i in range(len(d)):
    print(d[i]+" ", end="")
```

```python
#Piling Up!
from collections import deque

T = int(input())

for _ in range(T):
    n = int(input())
    kocke = deque(map(int, input().split()))

    prethodna = float("inf")
    moguce = True

    while kocke:
        if kocke[0] >= kocke[-1]:
            trenutna = kocke.popleft()
        else:
            trenutna = kocke.pop()

        if trenutna > prethodna:
            moguce = False
            break
        prethodna = trenutna

    print("Yes" if moguce else "No")
```

```python
#Map and Lambda Function
cube = lambda x: x**3

def fibonacci(n):
  a = []
  for i in range(n):
    if i==0:
      a.append(0)

    if i ==1:
      a.append(1)
    if i>1:
      x = a[i-1]+a[i-2]
      a.append(x)
  return a


if __name__ == '__main__':
    n = int(input())
    print(list(map(cube, fibonacci(n))))
```
